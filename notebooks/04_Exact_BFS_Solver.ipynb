```python
from collections import deque

initial_state = ("A", "B", "C", "D")
goal_state = ("B", "C", "D", "A")

# Path graph P4
# Python uses zero-based indexing
edges = [
    (0, 1),
    (1, 2),
    (2, 3)
]

print("Initial state:", initial_state)
print("Goal state:", goal_state)
print("Allowed SWAP edges:", edges)
```

```python
def get_successors(state, edges):
    successors = []

    for u, v in edges:
        next_state = list(state)

        next_state[u], next_state[v] = (
            next_state[v],
            next_state[u]
        )

        successors.append((tuple(next_state), (u, v)))

    return successors
```

```python
def bfs(initial, target, edges):
    queue = deque([initial])

    # States that have already been discovered
    visited = {initial}

    # Store the previous state of each discovered state
    parent = {initial: None}

    # Store the SWAP used to reach each state
    move_used = {}

    expanded_states = 0

    while queue:
        current = queue.popleft()
        expanded_states += 1

        # Check whether the goal has been reached
        if current == target:
            break

        # Generate all legal successor states
        for next_state, move in get_successors(current, edges):

            if next_state not in visited:
                visited.add(next_state)

                parent[next_state] = current
                move_used[next_state] = move

                queue.append(next_state)

    # No solution exists if target was never discovered
    if target not in parent:
        return None

    # Reconstruct the path from target to initial
    path = []
    current = target

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()

    # Reconstruct the SWAP sequence
    moves = []

    for state in path[1:]:
        moves.append(move_used[state])

    return {
        "path": path,
        "moves": moves,
        "visited_states": len(visited),
        "expanded_states": expanded_states
    }
```

```python
result = bfs(initial_state, goal_state, edges)

if result is None:
    print("No solution found.")

else:
    path = result["path"]
    moves = result["moves"]

    print("INITIAL:", path[0])
    print()

    for i, move in enumerate(moves):
        u, v = move

        print(f"SWAP({u+1}, {v+1})")
        print("   ->", path[i+1])

    print()
    print("GOAL:", path[-1])
    print("Minimum SWAP:", len(moves))
    print("Visited states:", result["visited_states"])
    print("Expanded states:", result["expanded_states"])
```

```python
assert result is not None

path = result["path"]
moves = result["moves"]

# The path must start at the initial state
assert path[0] == initial_state

# The path must end at the goal state
assert path[-1] == goal_state

# Each transition must correspond to exactly one SWAP
assert len(path) == len(moves) + 1

# Verify every recorded SWAP
for i, (u, v) in enumerate(moves):
    current = list(path[i])

    current[u], current[v] = current[v], current[u]

    assert tuple(current) == path[i+1]

print("All validation tests passed.")
```

```python
def inversion_count(state):
    count = 0

    for i in range(len(state)):
        for j in range(i + 1, len(state)):
            if state[i] > state[j]:
                count += 1

    return count


test_states = [
    ("A", "B", "C", "D"),
    ("B", "C", "D", "A"),
    ("D", "C", "B", "A")
]

for state in test_states:
    print(
        state,
        "-> Inversions:",
        inversion_count(state)
    )
```
