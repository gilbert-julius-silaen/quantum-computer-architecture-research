```python
from itertools import permutations
from math import factorial

for n in range(1, 8):
    qubits = list(range(1, n + 1))

    generated = len(list(permutations(qubits)))
    theoretical = factorial(n)

    print(
        f"n={n}: "
        f"Generated={generated}, "
        f"Theoretical={theoretical}"
    )

    assert generated == theoretical
```

```python
# Initial permutation
state = [1, 2, 3]

# Identity permutation
identity = [1, 2, 3]

print("Initial:", state)
print("Identity:", identity)

# Apply transposition (1 2)
state[0], state[1] = state[1], state[0]
print("After SWAP(1, 2):", state)

# Apply the same transposition again
state[0], state[1] = state[1], state[0]
print("After SWAP(1, 2) again:", state)

assert state == identity
```

```python
from itertools import combinations

elements = [1, 2, 3]

for u, v in combinations(elements, 2):
    state = elements.copy()

    # Python index starts from 0
    state[u-1], state[v-1] = state[v-1], state[u-1]

    print(f"Transposition ({u} {v}): {state}")
```
