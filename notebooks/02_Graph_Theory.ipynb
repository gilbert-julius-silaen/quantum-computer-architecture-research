```python
graph = {
    1: [2],
    2: [1, 3],
    3: [2, 4],
    4: [3]
}

print(graph)
```

```python
vertices = list(graph.keys())

edges = []

for vertex in graph:
    for neighbor in graph[vertex]:
        if vertex < neighbor:
            edges.append((vertex, neighbor))

print("Vertices:", vertices)
print("Number of vertices:", len(vertices))
print("Edges:", edges)
print("Number of edges:", len(edges))
```

```python
for vertex in graph:
    degree = len(graph[vertex])
    print(f"Degree of vertex {vertex}: {degree}")
```

```python
# Path graph P4
path_graph = {
    1: [2],
    2: [1, 3],
    3: [2, 4],
    4: [3]
}

# Cycle graph C4
cycle_graph = {
    1: [2, 4],
    2: [1, 3],
    3: [2, 4],
    4: [1, 3]
}

# 2x2 Grid graph
grid_graph = {
    1: [2, 3],
    2: [1, 4],
    3: [1, 4],
    4: [2, 3]
}

for name, topology in [
    ("Path P4", path_graph),
    ("Cycle C4", cycle_graph),
    ("Grid 2x2", grid_graph)
]:
    edge_count = sum(len(neighbors) for neighbors in topology.values()) // 2
    print(name, "->", edge_count, "edges")
```
