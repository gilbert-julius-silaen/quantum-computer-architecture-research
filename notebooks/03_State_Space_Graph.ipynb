```python
initial_state = ("A", "B", "C", "D")

edges = [
    (0, 1),
    (1, 2),
    (2, 3)
]

print("Initial state:", initial_state)
print("Allowed SWAP edges:", edges)
```

```python
def get_successors(state, edges):
    successors = []

    for u, v in edges:
        next_state = list(state)

        next_state[u], next_state[v] = (
            next_state[v],
            next_state[u]
        )

        successors.append(tuple(next_state))

    return successors
```

```python
successors = get_successors(initial_state, edges)

print("Initial state:", initial_state)
print("Successor states:")

for state in successors:
    print(state)
```

```python
from itertools import permutations

qubits = ["A", "B", "C", "D"]

all_states = list(permutations(qubits))

print("Total states:", len(all_states))
```

```python
state_graph = {}

for state in all_states:
    state_graph[state] = get_successors(state, edges)

print("State:", initial_state)
print("Connected states:")

for neighbor in state_graph[initial_state]:
    print(neighbor)
```

```python
# Count the number of states
number_of_states = len(state_graph)

# Count directed transitions
number_of_transitions = sum(
    len(neighbors)
    for neighbors in state_graph.values()
)

# Each undirected edge is counted twice
number_of_edges = number_of_transitions // 2

print("Number of states:", number_of_states)
print("Number of directed transitions:", number_of_transitions)
print("Number of undirected edges:", number_of_edges)
```
